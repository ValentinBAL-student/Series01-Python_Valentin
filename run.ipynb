```python

#import necessary libraries
import numpy as np
import matplotlib.pyplot as plt

#load data as ECGu.txt in the notebook from the data folder
data = np.loadtxt("data/ECGu.txt")
print(data)
#check the data shape
print(data.shape)
```

#plot Step 4.4 of the assignment.
The aim is to plot the raw data in the notebook using matplotlib library.
Check if the code is working and if the plot is correct.

```python
#plot the data by using matplotlib.pyplot and title it "ECG raw signal"
frequency = 1000  # Assuming a sampling rate of 1000 Hz

#define  function to plot the data
def plot_ecg(data, frequency,title="ECG raw Signal"):
    plt.plot(frequency, data)
    plt.title(title)
    plt.xlabel("frequency")
    plt.ylabel("Amplitude")
    plt.show()
```

```python
#plot the data by using matplotlib.pyplot and title it "ECG raw signal"
frequency = np.arange(0, len(data))

#define  function to plot the data
def plot_ecg(data, frequency,title="ECG raw Signal"):
    plt.plot(frequency, data)
    plt.title(title)
    plt.xlabel("frequency")
    plt.ylabel("Amplitude")
    plt.show()

plot_ecg(data, frequency)  # Call the function to plot the ECG signal
#when you define a function, remember to call it after defining it, otherwise it won't execute. 
```

Now we're starting to work on the data by using standard units, we will convert the data to it.
All of this, to plot the data in a more readable.
In the following cell we will convert using the "A/D converter gain"


```python
ecg_mV = data * 1024 #convert the data to mV by multiplying it by 1024µV
time_in_seconds = np.arange(0, len(data)) / frequency
plot_ecg(ecg_mV, time_in_seconds)  # Call the function to plot the ECG signal 
```
